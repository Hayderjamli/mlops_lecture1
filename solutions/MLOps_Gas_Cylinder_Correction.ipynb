# MLOps Architecture Design Exercise
## Gas Cylinder Consumption Forecasting

**Duration: 15–20 minutes**

### Business context

A company distributes gas cylinders across several depots.

Its objective is to anticipate consumption in order to have the right number of cylinders, in the right place, at the right time.

The company can:
- Reuse cylinders.
- Recycle cylinders and refill them with different gases.
- Transfer cylinders between depots.
- Manufacture new cylinders.

Poor forecasting leads to:
- **Underestimation:** stock shortages, low recycling rates and missed deliveries.
- **Overestimation:** unused inventory, unnecessary storage costs and inefficient resource allocation.

### Available data

The company has access to:
- Sales and delivery data.
- Inventory levels in each depot.
- Cylinder returns.
- Transaction dates.

Data is stored across several files and databases. Data quality varies and data volumes are increasing.

### Your mission

You are responsible for proposing an **MLOps architecture** to develop, deploy and maintain a machine learning model predicting gas cylinder consumption.

### Constraints

- Historical data is stored in a management system and can be exported to Amazon S3.
- Inventory management decisions are made during daily meetings.
- Several data scientists and data engineers collaborate on the code.
- Deployment should be automated as much as possible.
- Model performance must be monitored.
